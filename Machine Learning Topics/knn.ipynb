```python
import numpy as np 
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('breast-cancer.csv')
df.head()
```

```python
df.drop(columns=['id'], inplace=True)
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['diagnosis']), df.diagnosis, test_size=0.2, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

- important beacuse knn always depends on distance and vector, coz all values should be similar

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.fit_transform(X_test)
```

```python
# checking best k value for better accuracy score 
score = []

for i in range(1,31):
    knn = KNeighborsClassifier(n_neighbors=i)
    knn.fit(X_train, y_train)

    ypred = knn.predict(X_test)
    
    score.append(accuracy_score(y_test, ypred))
```

```python
import matplotlib.pyplot as plt

plt.plot(score)
```

```python
print(score[8])
print(max(score))
print(score)
```

```python
knn_f = KNeighborsClassifier(n_neighbors=9)
knn_f.fit(X_train, y_train)
```

```python
y_pred1 = knn_f.predict(X_test)
print(accuracy_score(y_test, y_pred1))
```

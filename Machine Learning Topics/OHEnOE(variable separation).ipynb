### One_Hot_Encoding (for nominal data) and Ordinal_Encoding (for ordinal data) using Column Transform 

```python
from sklearn.compose import ColumnTransformer
```

```python
# passing 3 Encoder in single columtransformer at ones 

transformer = ColumnTransformer(transformers=[
    ('tnf1', SimpleImputer(),['fever']),
    ('tnf2', OrdinalEncoder(categories=[['Mild', 'Strong']]),['cough']),
    ('tnf3',OneHotEncoder(sparse=False, drop='first'), ['gender', 'city'])
    ],remainder='passthrough')   # reminder : what to do with remaining columns "passthrough", or "drop"
```

```python
import subprocess
```

```python
# with shell 
subprocess.run('ls -la', shell=True)
```

```python
# without shell (should be in list)
subprocess.run(['ls', '-la'])
```

```python
# we can input code by storing it into variable
x = subprocess.run('ls -la', shell=True) 
print(x.args)
```

```python
# errors check -> 0 for no error, and 1
print(x.returncode)
```

```python
# to capture outputs (by default it will not store outputs)
y = subprocess.run('ls -la',capture_output=True ,shell=True)
# no new lines 
print(f'output : {y.stdout} \n')
# new lines
print(f'output \n : {y.stdout.decode()}')
```

```python
# Redirect output to file 
with open('output.txt', 'w') as f:
    p1 = subprocess.run('ls -la', stdout=f, text=True, shell=True)
```

```python
# how to capture error in a variable 
err = subprocess.run('ls -la file.txt', capture_output=True , text=True, shell=True)
print(err.returncode)   # 0 means no error 
print(err.stderr)
```

```python
# for python to throw error : check = True
err = subprocess.run('ls -la file.txt', capture_output=True, text=True, shell=True, check=True)
print(err)
```

```python
# ignore error 
err = subprocess.run('ls -la file.txt',stderr=subprocess.DEVNULL, shell=True)
print(err)
```

```python
# change inputs (output of current to input of new)

# -> Example extracts the position of test word in text.txt 
ch = subprocess.run('cat text.txt', capture_output=True, shell=True, text=True)
# print(ch.stdout)
ch2 = subprocess.run('grep -n test', capture_output=True, shell=True, input=ch.stdout, text=True)
print(ch2.stdout)
```
